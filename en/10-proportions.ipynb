```python
# count people by family size (FamilySize)
df.FamilySize.value_counts().sort_index()
```

```python
# plot the count of people by family size and class
_, ax = plt.subplots(figsize=(12,6))
_ = sns.countplot(data=df, x='FamilySize', hue='Pclass', ax=ax).set(
    title='Contagem de Pessoas por Qtd Familiares e Classe',
    xlabel='Tamanho da Família', ylabel='Qtd. Pessoas'
)
```

# 💡 Technical Tip!!

> Your first analyses sometimes explain nothing...
>
> `Solution`: Look for other ways to analyze. E.g. Show proportions (%).

<img src="https://media2.giphy.com/media/CiOHO5544doY/200.gif" width=500>

> Now I get it!

```python
sns.light_palette("blue", as_cmap=True)
```

```python
# create a table with the proportion of people in each class by family size
# and add conditional formatting to highlight the distribution
cm = sns.light_palette("blue", as_cmap=True)
tmp = pd.crosstab(df.FamilySize, df.Pclass, normalize=0)
tmp.style.background_gradient(cmap = cm)
```

# 💡 Technical Tip!!

> **Need to compare categorical variables?**
>
> > Keep `pd.crosstab` in your arsenal! 🔫 (`pandas >= 0.23.0`)

<img src="https://media1.giphy.com/media/26gsg1pkYA7B7cnuM/giphy.gif" width=500>

> Secret Weapon!

```python
# plot this proportion as a bar chart
_ = tmp.plot(
    kind='bar', figsize=(12,7),
    title='Proporção de Pessoas por Qtd Familiares e Classe',
    xlabel='Tamanho da Família', ylabel='% Pessoas',
    stacked=True
)
```

```python
# create a table highlighting the 3rd quartile (75%)
tmp.style.highlight_quantile(q_left=0.75, axis=1)
```

# 💡 Technical Tip!!

> **Want to make a table easier to read?**
>
> > Use conditional formatting with these methods (`pandas >= 1.3.0`)
> > * `pandas.io.formats.style.Styler.highlight_null`
> > * `pandas.io.formats.style.Styler.highlight_max`
> > * `pandas.io.formats.style.Styler.highlight_min`
> > * `pandas.io.formats.style.Styler.highlight_between`
> > * `pandas.io.formats.style.Styler.highlight_quantile`
> > * `pandas.io.formats.style.Styler.background_gradient`
> > * `pandas.io.formats.style.Styler.text_gradient`
