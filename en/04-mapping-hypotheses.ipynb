### Business Goal

> Identify the factors that led some Titanic passengers to survive and others not.
>
> `Reason`: make future cruises safer.

### Notebook Goal

> 1. Understand the process of raising hypotheses.
> 2. Raise hypotheses for the Case


### 🗺️ Mapped Hypotheses

Passengers more likely to survive:
1. From higher classes (`Pclass` 1 > 2 > 3)
2. With more expensive tickets (`Fare`)
3. With distinguished titles (`Name`)
4. Aged 20-35 (`Age`)
5. Female (`Sex`)
6. With small families (`Parch` & `SibSp`)

Bonus:
7. Larger families are from lower classes
8. Higher classes have older passengers

# 💡 Technical Tip!!

It is a cyclical process....

> Generating and validating hypotheses is an iterative process.
>
> As we explore the data, more hypotheses are generated.
>
> The secret is to stay in close contact with the business team.
>
> `Reason`: Make sure explorations are prioritized _correctly_!
