# Validating Hypothesis #2
> `Hypothesis`: Passengers with more expensive tickets (`Fare`) are more likely to survive

```python
# show the mean of `Survived`
df['SurvivedDescription'] = df['Survived'].replace({0: 'Não Sobreviveu', 1: 'Sobreviveu'})
df.groupby('SurvivedDescription').Fare.mean()
```

# 💡 Technical Tip!!

> Be careful with means!
>
> They can be misleading.
> 
> `Reason`: The mean for `Survived` may be higher because a few points are very high (outliers). While the rest of the distribution behaves completely differently. Maybe even lower than `Did Not Survive`.
>
> `Solution`: Visualize the full distribution.

<img src="https://i.imgur.com/t0dvgrD.gif" width=600/>

> *Be careful with means!*

# Let's take a closer look!

Histograms, box plots and KDE are great for exactly this:

> What the distribution of the data looks like!

<img src="https://media2.giphy.com/media/3oriO8vwmRIZO6kcNO/200.gif" width=600/>

> Is it really so?


```python
# plot the distribution of ticket fares
_, ax = plt.subplots(figsize=(12,7))
_ = sns.histplot(data=df, x='Fare', ax=ax, bins=50).set(
    title='Distribuição de Valor de Passagem',
    xlabel='Fare (Valor da Passagem)', ylabel='Quantidade de Pessoas'
)
```

```python
# plot the distribution of ticket fares, showing whether the passenger survived or not
_, ax = plt.subplots(figsize=(12,7))
_ = sns.histplot(data=df, x='Fare', hue='Survived', ax=ax, kde=True,
                 bins=50, multiple="dodge", shrink=0.75).set(
    title='Distribuição de Valor de Passagem por Sobreviventes',
    xlabel='Fare (Valor da Passagem)', ylabel='Quantidade de Pessoas'
)
```

# Let's start with `boxplot`?

The `boxplot` is also called (but almost never) a box chart.

It shows the data in boxes divided by lines.

The lines at the 1st, 2nd and 3rd quartile. They split the data at 25, 50 and 75%.

Remember that the 2nd quartile (50%) is the median.

The max and min lines are drawn as a multiple of the quartiles.

What lies beyond is an `outlier`, or behaves outside the norm.

<img src="https://c.tenor.com/DbBoAfOWRlAAAAAC/cats-cat.gif" width=500/>

> *My new car... Vroommm....*

```python
# visualize the distribution of `Fare` by `Survived` - boxplot
_, ax = plt.subplots(figsize=(7,7))
_ = sns.boxplot(data=df, x='SurvivedDescription', y='Fare', ax=ax).set(
    title='Distribuição de Valor de Passagem por Sobreviventes',
    ylabel='Fare (Valor da Passagem)', xlabel='Condição do Passageiro'
)
```

# We are on the right track...

Looking at the distributions, it really seems that those who survived paid a higher fare for the ticket...

Shall we draw a few more charts to confirm?

# How about trying the `violinplot`?

The `violinplot` shows the data as a density function.

Usually through the `kde` method (kernel density estimate).

Which is a fancy way of saying:

> Smooth the data with a nice curve. Instead of a heap of points.

<img src="https://media4.giphy.com/media/jlqJieWthmFZSpsHhe/giphy.gif" width=500/>


```python
# visualize the distribution of `Fare` by `Survived` - violinplot
_, ax = plt.subplots(figsize=(7,7))
_ = sns.violinplot(data=df, y='SurvivedDescription', x='Fare', ax=ax).set(
    title='Distribuição de Valor de Passagem por Sobreviventes',
    xlabel='Fare (Valor da Passagem)', ylabel='Condição do Passageiro'
)
```

```python
# validate statistically
compare2groups(
    df[df.Survived==1].Fare,
    df[df.Survived==0].Fare
    )
```

# 💡 Technical Tip!!

> Always start with the simplest!
>
> To compare distributions of numerical variables:
>
> 1. Start with the mean
> 2. Then make a `boxplot` and find the outliers
> 3. If needed, make a violinplot
