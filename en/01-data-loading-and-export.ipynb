### Import Libs

```python
import io, os, requests
import pandas as pd
```

### Function Declarations

```python
# create a method to read text from a url
def load_txt_from_url(url:str):
  """
  Carrega texto disponível em URL
  """
  return requests.get(url).text

# create a method to read a DataFrame from a URL
def load_df_from_url(url:str, sep:str=','):
  """
  Carrega dataframe disponível em URL (link externo)
  """
  with io.StringIO(load_txt_from_url(url)) as data:
    df = pd.read_csv(data, sep=sep)
  return df
```

### Declare constants

```python
TEST_URL  = 'https://raw.githubusercontent.com/murilobellatini/datasets/master/titanic/test.csv'
TRAIN_URL = 'https://raw.githubusercontent.com/murilobellatini/datasets/master/titanic/train.csv'
DICT_URL  = 'https://raw.githubusercontent.com/murilobellatini/datasets/master/titanic/data_dictionary.txt'
```

### Show the data dictionary

```python
# show the text content of `DICT_URL`
print(load_txt_from_url(DICT_URL))
```

### Load test data

```python
# load the dataframe from `TEST_URL` into variable `test`
test = load_df_from_url(TEST_URL)
test
```

### Show training data

```python
# load the dataframe from `TRAIN_URL` into variable `train`
train = load_df_from_url(TRAIN_URL)
train
```

### Export data

```python
# mount google drive
import os
from google.colab import drive
drive.mount('/content/gdrive')

# create a folder to store raw data
outdir = '/content/gdrive/MyDrive/BigDataDoctor/titanic/data/raw'
if not os.path.exists(outdir):
    os.makedirs(outdir)

# export DataFrames as CSV
test.to_csv(f'{outdir}/test.csv')
train.to_csv(f'{outdir}/train.csv')
```
