### Hypothesis

#### 📖 **What is it?**

* It is a proposed explanation for a phenomenon.
* It must be possible to validate (or test) it.

In short:

> In general, it is any hunch that needs more information to be confirmed.

#### ⚙️ **What is it for?**

* To start any investigation (scientific or business).
* It guides (almost) every decision
* Why?
  * The information available is (almost) always limited!

<img src="https://media4.giphy.com/media/24akSucLOFwwoZamdr/200.gif" width="600"/>

> *It is very important!*


<br/>
<br/>




#### 🤷 **Why should I care?**

* The role of the Data Analyst:
  * Give the business team information to make better decisions (data-driven).
  * Validate (or invalidate) with data.

In short:

> To do well in Data, it is essential to guide all your analyses with hypotheses.
>
> And more than that, to get good at validating them.
