# Validating Hypothesis #5

Do women have a better chance of survival?

<img src="https://media2.giphy.com/media/3o7btXgbfRrJWlQRJm/200.gif" width=500>

> *Yayyyy... Ouch!*

```python
# show the survival rate by sex
df.groupby('Sex').Survived.mean()
```

```python
# compare whether there is a statistical difference between the samples
compare2groups(
    df[df.Sex=='male'].Survived.dropna(),
    df[df.Sex=='female'].Survived.dropna()
    )
```

# 👌 Insight / Conclusion #1

Hypothesis #5 validated. ✔️

Women have a much higher chance of survival.

# ⚡ Feature Engineering Express!!!


<img src="https://c.tenor.com/a9BzZfLW20YAAAAC/the-flash-barry-allen.gif" width=500>

```python
df['FamilySize'] = df.Parch + df.SibSp + 1
df.FamilySize.value_counts().sort_index()
```

# 💡 Technical Tip!!

> the_convention_in_python_is_snake_case
>
> ButSinceTheDatasetIsInCamelCaseItIsBetterToStayConsistent
>
> `That is`:
>
> > Whenever you can, use `snake_case`!
> >
> > Otherwise go with `CamelCase`.
>
> `Reason`
>
> > 1. Snake Case is easier to read
> > 2. It is the Python convention

# 👌 Insight / Conclusion #2

Hypothesis #6 rejected.

> Families of 4 members were the ones that survived the most.

But there is no predominance of survivors among small families (1 to 3 members).
