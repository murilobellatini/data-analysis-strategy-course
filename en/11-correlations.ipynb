# ⚡ Hypothesis Express!!

We did not compare two numerical variables...

> Can we use the ticket price (`Fare`) as a proxy for the class (`Pclass`)?

<img src="https://media0.giphy.com/media/6q0qelNtLoas4025sy/giphy.gif" width=500>

> Eureka!

```python
# compute Fare statistics grouped by Pclass
df.groupby('Pclass').Fare.describe()
```

```python
# visualize a violinplot with the distribution of Fare by Pclass and Sex
_, ax = plt.subplots(figsize=(12,10))
_ = sns.violinplot(
    data=df, y='Fare', x='Pclass',hue='Sex', split=True, ax=ax).set(
    title='Distribuição de Valor de Passagem por Sexo e Classe',
    xlabel='Classe (Pclass)', ylabel='Valor da Passagem (Fare)'
)
```

```python
# visualize a boxplot with the distribution of Fare by Pclass and Sex
_, ax = plt.subplots(figsize=(12,10))
_ = sns.boxplot(
    data=df, y='Fare', x='Pclass',hue='Sex', ax=ax).set(
    title='Distribuição de Valor de Passagem por Sexo e Classe',
    xlabel='Classe (Pclass)', ylabel='Valor da Passagem (Fare)'
)
```

# 👌 Insight / Conclusion #1

Hypothesis Express validated. ✔️

> Higher classes have a more expensive ticket price.

## Correlation Coefficients

There are mathematical methods to calculate correlation.

**But what does that mean?**

> It is how strongly two variables "go together" (or vary together), that is, how they correlate.

**Example**

> Weight and Height... Taller people tend to be heavier. It is not a rule, but there is a positive correlation between these variables.

<img src="https://c.tenor.com/oPT-PWAwAGUAAAAM/jose-anthony.gif" width=300>

> *Taller is heavier, shorter is lighter :)*

### Types of Correlation

We can find three types of correlation, as below:

1. Positive Correlation
2. No Correlation
3. Negative Correlation

<img src="https://www.statisticshowto.com/wp-content/uploads/2012/10/pearson-2-small.png" width=500>

### Methods

There are several methods to calculate correlation. One of the most famous is Pearson (the same guy as the p-value).

If you are interested in the formulas and differences, leave a comment!

```python
# compute the correlation between Age and Fare
df.Age.corr(df.Fare)
```

```python
# compute the correlation between Age and Fare by Pclass
for c in range(1,4):
  mask = (df.Pclass == c)
  corr = df[mask].Age.corr(df[mask].Fare)
  print(f'Pclass {c} Correlation: {corr}')
```

# 💡 Technical Tip!!

Convention for data (and academia).

> A coefficient between 0.5 and 1 ( or -1 and -0.5 ) indicates a strong correlation.
