# 💡 Technical Tip!!

> When in doubt about an insight, segment!
>
> `How?`
>
> Bucketing! Let's split the data into buckets, or `bins`...

<img src="https://media3.giphy.com/media/l3diH7dDO7f897ZLi/giphy.gif" width=500>

# pd.cut

> Creates buckets (data ranges) by splitting the values into equal parts.

```python
pd.cut(df.Age, bins=8).value_counts().sort_index()
```

```python
pd.cut(df.Age, bins=8)
```

```python
df.Age
```

# pd.qcut

> Creates buckets by splitting into equal quantities (the resulting counts are equivalent).
>
> With `q=4` the split corresponds to quartiles.

```python
pd.qcut(df.Age, q=8).value_counts().sort_index()
```

# But why is the data unbalanced?

> Look, there are 100 items in the 1st range, and 79 in the second...

<img src="https://c.tenor.com/NjhGnn9xk7UAAAAM/baby-scale.gif" width=500>

```python
df.Age.value_counts().sort_index().head(30)
```

> Because in the 16-year range there are 17 people, who go into the 1st range.

```python
df['Age_bins'] = pd.qcut(df.Age, q=8)
tmp = df.groupby('Age_bins').Survived.mean().sort_index()
tmp
```

# Let's visualize this data?

<img src="https://media3.giphy.com/media/lHoS3eSVSTcIw/giphy.gif" width=500>

> *Look at that!*

```python
_ = (tmp*100).plot(
    kind='bar', figsize=(12,6),
    title='Taxa de Sobrevivência por Faixa Etária',
    ylabel='% Sobreviventes', xlabel='Faixas Etárias'
    )
```

```python
compare2groups(
    df[(df.Age_bins==pd.Interval(0.419, 16.0)) & (df.Survived==1)].Age,
    df[(df.Age_bins==pd.Interval(16.0, 20.125)) & (df.Survived==1)].Age   
)
```

# 👌 Insight / Conclusion

Hypothesis #4 rejected. ❌

> There is a higher proportion of young survivors (up to 16 years).
>
> But outside the 20-35 age range.
