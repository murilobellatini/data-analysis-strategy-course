```python
# dividir "Name" em palavras (ou entre seus espaços)
df.Name.str.split(" ")
```

```python
# amazenar palavras de name em coluna "Name_list"
df['Name_list'] = df.Name.str.split()

# explodir coluna "Name_list" como Tidy Data
df.explode('Name_list')
```

# 💡 Sacada Técnica!!

> `Tidy Data` é incrível!
>
> Acabamos de reestruturar nossos dados neste formato.
>
> Eles seguem a tríade 3️⃣ abaixo:
>
> 1. Cada coluna é uma varíavel
> 2. Cada linha é uma observação
> 3. Cada célula tem um valor único
> 
> `Motivo`: Muito mais fácil analisar os dados e extrair métricas.
>
> ⚠️ `Alerta`: Consome mais memória/armazenamento por ser um formato denormalizado (dados redundantes).

<img src="https://i.pinimg.com/originals/41/5b/fd/415bfd2c979b0f4c671a3a133e441785.gif" width=500/>

> *Deixando a casa __tidy__ (organizada)!* 

```python
# contar palavras da coluna "Name_list"
df.Name_list.explode().value_counts()
```

```python
# mostrar 20 palavras mais recorrentes da coluna "Name_list"
df.Name_list.explode().value_counts().head(20)
```

```python
# guadar palavras da coluna "Name_list" em variável `tmp`
tmp = df.Name_list.explode()
tmp
```

```python
# contar palavras da variável `tmp` que contenham ponto "."
mask = tmp.str.contains('\.')
tmp[mask].value_counts()
```

```python
# extrair da coluna "Name" palavras que terminam com ponto "."
# e guardar em coluna "Title"
df['Title'] = df.Name.str.extract('(\w*\.)')
df.head()
```

# 💡 Sacada Técnica!!

> Acabamos de montar uma Expressão Regular (ou RegEx).
>
> Elas foram feitas para manipular textos.
>
> `Motivo`:

>
> ⚠️ `Alerta`:

```python
# contar valores de coluna "Title"
df.Title.value_counts()
```

```python
# remover coluna "Name_list"
df = df.drop('Name_list', axis=1)
```

# Bora enxugar essa lista?

<img src="https://ksr-ugc.imgix.net/assets/028/543/333/22d975b407a492ab0222c5febac6d607_original.gif?ixlib=rb-4.0.2&w=680&fit=max&v=1585256901&auto=format&gif-q=50&q=92&s=cb6d034ecbb2282bcfc68fcbb85f17a5" width=500>

```python
# selecionar em lista/array todos os títulos da lista, fora os 5 primeiros (top 4)
uncommon_titles = df.Title.value_counts().index[4:].values
uncommon_titles
```

```python
#  filtrar `df` só títulos incomuns
mask = df.Title.isin(uncommon_titles)
df[mask].Title.value_counts()
```

```python
# substituir títulos incomuns por "Uncommon Title"
# e contar novos títulos 
df.loc[mask, "Title"] = 'Uncommon Title'
df.Title.value_counts()
```

# Validação da Hipótese #3

> `Hipótese`: Tem maior probabilidade de sobreviver os passageiros com títulos de renome (`Name`)

Com os dados preparados, vamos para ação!

```python
# gerar percentual de sobreviventes por título
survived_title = df.groupby('Title').Survived.mean().rename('Survived_perc').sort_values(ascending=False)
survived_title 
```

```python
# gerar estatísticas (contagem, média, desvio padrão, etc) de sobreviventes por título
tmp = df.groupby('Title').Survived.describe().sort_values('mean', ascending=False)
tmp
```

# Vamos deixar mais visual?

> Para isso o histograma é perfeito!

> Ele serve para apresentar a distribuição dos seus dados categóricos.

```python
# visualizar os títulos como histograma
_ = sns.histplot(data=df, y='Title')
```

```python
# gerar gráfico barra com médias de sobreviventes
_ = (100*tmp['mean']).plot(
    kind='bar', table=True, legend=False,
    title='Distribuição de Sobreviventes por Título',
    ylabel='% Sobreviventes', xlabel='\n\nTítulo do Passageiro',
    fontsize=14, xticks=(),
    figsize=(14,7)
    )
```

```python
# apresentar titulos por sexo
df.groupby(['Sex', 'Title']).size()
```

```python
# validar estatísticamente
compare2groups(
    df[df.Title=='Mrs.'].Survived,
    df[df.Title=='Miss.'].Survived
    )
```

```python
# validar estatísticamente
compare2groups(
    df[df.Title=='Mrs.'].Survived,
    df[df.Title=='Master.'].Survived
    )
```
