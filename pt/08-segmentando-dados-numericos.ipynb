# 💡 Sacada Técnica!!

> Na dúvida sobre um insight, faça uma segmentação!

> Bucketização! Vamos dividir os dados em baldes, ou `bins`...

# pd.cut

> Cria buckets (faixas de dados) quebrando os valores em partes iguais.

```python
pd.cut(df.Age, bins=8).value_counts().sort_index()
```

```python
pd.cut(df.Age, bins=8)
```

```python
df.Age
```

# pd.qcut

> Cria buckets quebrando em quantidades iguais (a contagem resultado é equivalente).
>
> Com `q=4` a divisão corresponde em quartis.

```python
pd.qcut(df.Age, q=8).value_counts().sort_index()
```

# Mas por que os dados estão desbalanceados?

> Olha aí, tem 100 items na 1ª faixa, e 79 na segunda...

<img src="https://c.tenor.com/NjhGnn9xk7UAAAAM/baby-scale.gif" width=500>

```python
df.Age.value_counts().sort_index().head(30)
```

> Porque na faixa de 16 anos tem 17 pessoas, que entram na 1ª faixa.

```python
df['Age_bins'] = pd.qcut(df.Age, q=8)
tmp = df.groupby('Age_bins').Survived.mean().sort_index()
tmp
```

# Vamos visualizar estes dados?

<img src="https://media3.giphy.com/media/lHoS3eSVSTcIw/giphy.gif" width=500>

> *Olha só!*

```python
_ = (tmp*100).plot(
    kind='bar', figsize=(12,6),
    title='Taxa de Sobrevivência por Faixa Etária',
    ylabel='% Sobreviventes', xlabel='Faixas Etárias'
    )
```

```python
compare2groups(
    df[(df.Age_bins==pd.Interval(0.419, 16.0)) & (df.Survived==1)].Age,
    df[(df.Age_bins==pd.Interval(16.0, 20.125)) & (df.Survived==1)].Age   
)
```

# 👌 Insight / Conclusão

Hipótese #4 rejeitada. ❌

> Existe uma proporção maior de sobreviventes jovens (até 16 anos).
>
> Porém fora da faixa de 20-35 anos.
