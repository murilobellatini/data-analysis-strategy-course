# Validação da Hipótese #2
> `Hipótese`: Tem maior probabilidade de sobreviver os passageiros com tickets mais caros (`Fare`)

```python
# mostrar média de `Survived`
df['SurvivedDescription'] = df['Survived'].replace({0: 'Não Sobreviveu', 1: 'Sobreviveu'})
df.groupby('SurvivedDescription').Fare.mean()
```

# Vamos avaliar melhor!

Histogramas, gráficos caixa e KDE são ótimos para avaliar justamente isto:

> Como é a distribuição dos dados!

```python
# plotar distribuição de valor de passagem
_, ax = plt.subplots(figsize=(12,7))
_ = sns.histplot(data=df, x='Fare', ax=ax, bins=50).set(
    title='Distribuição de Valor de Passagem',
    xlabel='Fare (Valor da Passagem)', ylabel='Quantidade de Pessoas'
)
```

```python
# plotar distribuição de valor de passagem visualizando se sobreviveu ou não
_, ax = plt.subplots(figsize=(12,7))
_ = sns.histplot(data=df, x='Fare', hue='Survived', ax=ax, kde=True,
                 bins=50, multiple="dodge", shrink=0.75).set(
    title='Distribuição de Valor de Passagem por Sobreviventes',
    xlabel='Fare (Valor da Passagem)', ylabel='Quantidade de Pessoas'
)
```

# Vamos começar com `boxplot`?

O `boxplot`, também é (mas quase nunca) chamado de gráfico caixa.

Ele apresenta os dados em caixas dividas por linhas.

As do 1º, 2º e 3º quartil. Elas representam a divisão dos 25, 50 e 75% dos dados.

Lembrando que o 2º quartil (50%) diz respeito à mediana.

As linhas de máximo é mínimo são traçadas como um múltimo dos quartis.

O que fica além é `outlier`, ou tem um comportamento fora do padrão.

<img src="https://c.tenor.com/DbBoAfOWRlAAAAAC/cats-cat.gif" width=500/>

> *Meu carro novo... Vroommm....* 

```python
# visualizar distribuição de `Fare` por `Survived` - boxplot
_, ax = plt.subplots(figsize=(7,7))
_ = sns.boxplot(data=df, x='SurvivedDescription', y='Fare', ax=ax).set(
    title='Distribuição de Valor de Passagem por Sobreviventes',
    ylabel='Fare (Valor da Passagem)', xlabel='Condição do Passageiro'
)
```

# Estamos no caminho...

Avaliando as distribuições, realmente parece que quem sobreviveu pagou um tarifa maior no ticket...

# Que tal testar o `violinplot`?

O `violinplot` apresenta os dados como uma função de densidade.

Geralmente, pelo método `kde` (kernel densisty estimate).

Que é um jeito chique de dizer:

> Suavizar os dados com uma curva bonita. Ao invés de um amontoado de pontos.

<img src="https://media4.giphy.com/media/jlqJieWthmFZSpsHhe/giphy.gif" width=500/>


```python
# visualizar distribuição de `Fare` por `Survived` - violinplot
_, ax = plt.subplots(figsize=(7,7))
_ = sns.violinplot(data=df, y='SurvivedDescription', x='Fare', ax=ax).set(
    title='Distribuição de Valor de Passagem por Sobreviventes',
    xlabel='Fare (Valor da Passagem)', ylabel='Condição do Passageiro'
)
```

```python
# validar estatísticamente
compare2groups(
    df[df.Survived==1].Fare,
    df[df.Survived==0].Fare
    )
```
