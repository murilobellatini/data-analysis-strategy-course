# ⚡ Feature Engineering Express!!!


<img src="https://c.tenor.com/a9BzZfLW20YAAAAC/the-flash-barry-allen.gif" width=500>

```python
df['FamilySize'] = df.Parch + df.SibSp + 1
df.FamilySize.value_counts().sort_index()
```

# 💡 Sacada Técnica!!

> a_convenção_em_python_é_snake_case
>
> MasComoODatasetEstáEmCamelCaseMelhorÉManterConsistência
>
> `Ou seja`:
>
> > Sempre que puder utilize `snake_case`!
> >
> > Caso contrário vá para o `CamelCase`.
>
> `Motivo`
>
> > 1. Snake Case é mais fácil de ler
> > 2. É a convenção em Python
> 
> ⚠️ `Alerta`: Quebre a regra para manter consistência (mais fácil de ler)
>

```python
df
```

```python
tmp = df.groupby('FamilySize').Survived.mean()*100
_ = tmp.plot(kind='bar', figsize=(12,6),
             title='Taxa de Sobreviventes por Quantidade de Familiares',
             ylabel='% Sobreviventes', xlabel='Tamanho da Família')
```

# 👌 Insight / Conclusão #2

Hipótese #6 rejeitada.

> Famílias de 4 membros foram as que mais sobrevieram.

Porém, não há uma predominâncias de sobreviventes para famílias pequenas (1 a 3 membros).
