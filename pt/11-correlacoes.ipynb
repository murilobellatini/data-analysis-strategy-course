## Coeficientes de Correlação

Existem métodos matemáticos para calcular correlação.

**Mas o que isso quer dizer?**

> É a intensidade em que duas variáveis "andam juntas" (ou variam juntas), ou seja, como elas se correlacionam.

**Exemplo**

> Peso e Altura... Pessoas mais altas tendem a ser mais pesadas. Não é uma regra, porém existe uma correlação postiivas entre estas variáveis.

<img src="https://c.tenor.com/oPT-PWAwAGUAAAAM/jose-anthony.gif" width=300>

> *Maior é mais pesado, menor mais leve :)*

### Tipos de Correlação

Podemos encontrar três tipos de correlação como abaixo:

1. Correlação Positiva
2. Sem Correlação
3. Correlação Negativa

<img src="https://www.statisticshowto.com/wp-content/uploads/2012/10/pearson-2-small.png" width=500>

### Métodos

Existem diversos métodos para calcular correlação. Um dos mais famosos é o Pearson (o mesmo cara do p-value).

Se tiver interesse nas fórmulas e diferenças, comenta aí!

```python
# calcular correção entre Age e Fare
df.Age.corr(df.Fare)
```

```python
# calcular correção entre Age e Fare por Pclass
for c in range(1,4):
  mask = (df.Pclass == c)
  corr = df[mask].Age.corr(df[mask].Fare)
  print(f'Pclass {c} Correlation: {corr}')
```

# 💡 Sacada Técnica!!

Convenção para dados (e meio acadêmico).

> Coeficiente entre 0.5 e 1 ( ou -1 e -0.5 ) indica uma forte correlação.
